```python
import pandas as pd
```

```python
import duckdb
```

```python
df = pd.read_csv("Steel_industry_data.csv")
```

```python
duckdb.sql("""
SELECT *
From df
LIMIT 10
""").df()
```

```python
duckdb.sql("""
SELECT
    Usage_kWh,
    Load_type,
    Day_of_week
FROM df
WHERE Usage_kWh > 100
ORDER BY Usage_kWh DESC
""").df()
```

```python
duckdb.sql("""
SELECT *
FROM df
WHERE Load_Type = 'Medium_Load'
OR Load_Type = 'Maximum_Load'
ORDER BY Usage_kWh DESC
""").df()
```

```python
df["Load_Type"].unique()
```

```python
duckdb.sql("""
SELECT 
    Load_Type,
    AVG(Usage_kWh) AS avg_usage_kwh
FROM df
GROUP BY Load_Type
ORDER BY avg_usage_kwh DESC
""").df()
```

```python
duckdb.sql("""
SELECT
    Load_Type,
    COUNT(*) AS n_observations,
    AVG(Usage_kWh) AS avg_usage_kwh,
    MIN(Usage_kWh) AS min_usage_kwh,
    MAX(Usage_kWh) AS max_usage_kwh
FROM df
GROUP BY Load_Type
ORDER BY avg_usage_kwh DESC
""").df()
```

```python
duckdb.sql("""
SELECT 
    Day_of_week,
    AVG(Usage_kWh) as avg_usage_kwh
FROM df
GROUP BY Day_of_week
ORDER BY avg_usage_kwh DESC
""").df()
```

```python
duckdb.sql("""
SELECT
    WeekStatus,
    COUNT(*) AS n_observations,
    AVG(Usage_kWh) AS avg_usage_kwh
FROM df
GROUP BY WeekStatus
ORDER BY avg_usage_kwh DESC
""").df()
```

```python
duckdb.sql("""
SELECT
    Load_Type,
    AVG(Usage_kWh) AS avg_usage_kwh
FROM df
GROUP BY Load_Type
HAVING AVG(Usage_kWh) > 50
ORDER BY avg_usage_kwh DESC
""").df()
```

```python
duckdb.sql("""
SELECT
    CASE
        WHEN Usage_kWh < 25 THEN 'Low'
        WHEN Usage_kWh < 75 THEN 'Medium'
        ELSE 'High'
    END AS usage_category,
    AVG(Usage_kWh) as avg_usage_kwh,
    COUNT(*) AS observations
FROM df
GROUP BY usage_category
ORDER BY avg_usage_kwh
""").df()
```

```python
duckdb.sql("""
SELECT
    Load_Type,
    AVG(Usage_kWh) as avg_usage_kwh
FROM df
GROUP BY Load_Type
HAVING AVG(Usage_kWh) > 50
ORDER BY avg_usage_kwh DESC
""").df()
```

```python
Load_info = pd.DataFrame({
    "Load_Type": [
        'Light_Load',
        'Medium_Load',
        'Maximum_Load'
    ],
    "Load_Priority": [
        1,
        2,
        3
    ]
})
```

```python
duckdb.sql("""
SELECT
    df.Usage_kWh,
    df.Load_Type,
    Load_info.Load_Priority
FROM df
INNER JOIN Load_info
    ON df.Load_Type = Load_info.Load_Type
LIMIT 20
""").df()
```

```python
duckdb.sql("""
SELECT
    df.Usage_kWh,
    df.Load_Type,
    Load_info.Load_Priority
FROM df
LEFT JOIN Load_info
    ON df.Load_Type = Load_info.Load_Type
LIMIT 20
""").df()
```

```python
duckdb.sql("""
WITH load_summary AS (
    SELECT
        Load_Type,
        AVG(Usage_kWh) AS avg_usage_kwh,
        COUNT(*) AS observations
    FROM df
    GROUP BY Load_Type
)

SELECT *
FROM load_summary
WHERE avg_usage_kwh > 50
ORDER BY avg_usage_kwh DESC
""").df()
```

```python
duckdb.sql("""
SELECT 
    Load_Type,
    Usage_kWh,
    AVG(Usage_kWh) OVER (
        PARTITION By Load_Type
    ) AS load_type_avg
FROM df
LIMIT 20
""").df()
```

```python
duckdb.sql("""
SELECT 
    Load_Type,
    Usage_kWh,
    ROW_NUMBER() OVER(
        PARTITION BY Load_Type
        ORDER BY Usage_kWh DESC
    ) AS usage_rank
FROM df
""").df()
```

```python
con = duckdb.connect("steel_energy.duckdb")
```

```python
con.sql("""
CREATE OR REPLACE TABLE energy_usage AS
SELECT *
FROM df
""")
```

```python
con.sql("""
SELECT
    Load_Type,
    AVG(Usage_kWh) AS avg_usage_kwh
FROM energy_usage
GROUP BY Load_Type
ORDER BY avg_usage_kwh DESC
""").df()
```
